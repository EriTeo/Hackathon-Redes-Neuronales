# Spaceship Titanic - Machine Learning
Este notebook carga los datos, los preprocesa, entrena un modelo Random Forest y genera las predicciones.

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder

# Cargar los datos (Hay que subir los archivos train.csv y test.csv a Colab primero)
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
test_ids = test_df['PassengerId']
```

```python
# Preprocesamiento
def preprocess_data(df):
    df = df.copy()
    df[['Cabin_Deck', 'Cabin_Num', 'Cabin_Side']] = df['Cabin'].str.split('/', expand=True)
    df = df.drop(['Cabin', 'Name', 'PassengerId'], axis=1)

    num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck', 'Cabin_Num']
    cat_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'VIP', 'Cabin_Deck', 'Cabin_Side']

    num_imputer = SimpleImputer(strategy='median')
    cat_imputer = SimpleImputer(strategy='most_frequent')

    df[num_cols] = num_imputer.fit_transform(df[num_cols])
    df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

    for col in cat_cols:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))

    return df

X = preprocess_data(train_df.drop('Transported', axis=1))
y = train_df['Transported']
X_test = preprocess_data(test_df)
```

```python
# Entrenar modelo
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

# Predecir
predictions = model.predict(X_test)

# Generar y guardar archivo
submission = pd.DataFrame({
    'PassengerId': test_ids,
    'Transported': predictions
})
submission.to_csv('sample_submission.csv', index=False)
print("Archivo 'sample_submission.csv' generado exitosamente.")
submission.head()
```
